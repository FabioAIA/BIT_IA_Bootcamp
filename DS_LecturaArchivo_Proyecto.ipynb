Informacion contenida: Colombianos registrados en el exterior

Fuente: Ministerio de Relaciones exteriores

Desripcion de las columnas:

Código ISO país	ISO del país de residencia. Cuando no se tienen datos registrados se presenta el texto (NO REGISTRA)	c_digo_iso_pa_s	Texto

Ciudad de Residencia	Ciudad de residencia del inscrito, Departamento/Ciudad	ciudad_de_residencia	Texto

Oficina de registro	Consulado donde la persona realizó su registro consular. Cuando no se tienen datos registrados se presenta el texto (NO REGISTRA)	oficina_de_registro	Texto

Grupo edad	El grupo se basa en la edad de acuerdo a lo siguiente: Entre 0 y 5 años Primera infancia Entre 6 y 11 años Infante Entre 12 y 18 años Adolescente Entre 19 y 28 años Adulto joven Entre 29 y 59 años Adulto Entre 60 en adelante Adulto Mayor.
Leer más	grupo_edad	Texto

Edad (años)	Edad de la persona. Cuando no se tienen datos registrados se el valor -1.	edad_a_os	Número

Área Conocimiento	Cuando no se tienen datos registrados se presenta el texto (NO REGISTRA)	rea_conocimiento	Texto

Sub Area Conocimiento	Cuando no se tienen datos registrados se presenta el texto (NO REGISTRA)	sub_area_conocimiento	Texto

Nivel Académico	Cuando no se tienen datos registrados se presenta el texto (NO REGISTRA)	nivel_acad_mico	Texto

Estado civil	Cuando no se tienen datos registrados se presenta el texto (NO REGISTRA)	estado_civil	Texto

Género	Cuando no se tienen datos registrados se presenta el texto (DESCONOCIDO)	g_nero	Texto

Etnia de la persona	Cuando no se tienen datos registrados se presenta el texto (NO REGISTRA)	etnia_de_la_persona	Texto

Estatura (CM)	Estatura dada en centímetros, el valor -1 indica que no se encuentra una estatura registrada.	estatura_cm	Número

Ciudad de Nacimiento	Ciudad de Nacimiento del inscrito	ciudad_de_nacimiento	Texto

Localización	Localización del ciudadano georeferenciada según el país de residencia	localizaci_n	Ubicación

 Registro	Fecha de Registro en formato año-mes	fecha_de_registro	Texto

Cantidad de personas	Cantidad de personas residentes en el exterior que cumplen con las condiciones demográficas	cantidad_de_personas	Número
